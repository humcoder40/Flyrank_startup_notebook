Recall (finding every single decaying page) is completely irrelevant if the backlog is 10,000 pages long. We need the model's top 50 recommendations to be highly accurate, actionable candidates.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

Why ML > Fixed Rules:
A fixed heuristic rule (like "age > 180 days and trend == down") creates massive, unprioritized ties. As shown in our initial data exploration, simple rules flagged nearly 10,000 pages. A machine learning model beats this by learning the complex, non-linear interactions between feature weights—such as how the slope of CTR decay interacts with search volume and position volatility—allowing us to force-rank those 10,000 pages into a strict, mathematically justified priority queue.
